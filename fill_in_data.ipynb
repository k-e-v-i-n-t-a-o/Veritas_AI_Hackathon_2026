```python
from csv import reader

day3 = 5
AIRQUALITY3 = 10.09
AIRQUALITY2 = 10.35
AIRQUALITY1 = 8.14
RAINTIME3, WINDSPEED3 = [3.0,20.4]
RAINTIME2, WINDSPEED2 = [5.0,18.0]
RAINTIME1, WINDSPEED1 = [4.0,12.9]


day2 = (day3+1)%7
day1 = (day3+2)%7
```

awaken my model

```python
from pickle import load


with open("MyVeryOwnModelThatImVeryProudOvAndHelloIfUrReadingThis.pkl", mode='rb') as file:
    model = load(file)

with open("MyScalerForMyVeryOwnModelYayHappyFace", mode='rb') as file:
    scaler = load(file)
```

predict tomorrows air quality!

```python

values = [(
    AIRQUALITY3,
    day3,
    RAINTIME3,
    WINDSPEED3,
    AIRQUALITY2,
    day2,
    RAINTIME2,
    WINDSPEED2,
    AIRQUALITY1,
    day1,
    RAINTIME1,
    WINDSPEED1
)]

values = scaler.transform(values)

pred = model.predict(values)
print(pred)
```

Now lets use that prediction as the next day's air quality!

```python
with open("data/extra_data.csv", mode='r') as file:
    extra_data = list(reader(file))

values = [(
    AIRQUALITY2,
    day2,
    RAINTIME2,
    WINDSPEED2,
    AIRQUALITY1,
    day1,
    RAINTIME1,
    WINDSPEED1,
    float([pred][0][0]),
    (day3+3)%7,
    float(extra_data[0][-2]),
    float(extra_data[0][-1])
    )]

values = scaler.transform(values)
pred2 = model.predict(values)
```
