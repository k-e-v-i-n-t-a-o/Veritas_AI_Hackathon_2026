```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

data = None
values, answers = data.data, data.target
train_values, test_values, train_answers, test_answers = train_test_split(values, answers, test_size=0.2)

model = RandomForestClassifier()

#train the model
model.fit(train_values, train_answers)

#make predictions
preds = model.predict(test_values)

#evaluate model
accuracy = accuracy_score(test_answers, preds)
print(f"Accuracy: {accuracy:.4%}")
```
